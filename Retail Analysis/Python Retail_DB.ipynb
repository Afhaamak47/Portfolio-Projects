```python
#Extract file from zip files
import zipfile

# Path to the downloaded zip file
zip_file = "retail-orders.zip"

# Folder where you want to extract
extract_folder = "retail_orders"

with zipfile.ZipFile(zip_file, 'r') as zip_ref:
    zip_ref.extractall(extract_folder)

print(f"Files extracted to '{extract_folder}' folder")
```

```python
import pandas as pd

# Replace 'orders.csv' with the actual file name inside the zip
data = pd.read_csv("retail_orders/orders.csv", na_values = ["Not Available","unknown"]) 
# data.head(20)
data["Ship Mode"].unique()
```

```python
data
```

```python
# data.rename(columns = ({'Order Id':'order_id'}))
data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(' ','_')

data
```

```python
#Calculating discount
data["discount_price"] = data['list_price'] * data['discount_percent'] / 100
data["total_sales"] = data['list_price'] - data['discount_price']
data["profit"] = data['total_sales'] - data['cost_price']
data
```

```python
# Correcting the column data types

data['order_date'] = pd.to_datetime(data['order_date'], format = "%Y-%m-%d")
data.dtypes
```

```python
columns_to_drop = ['discount_percent', 'cost_price', 'list_price']
existing_cols_to_drop = [col for col in columns_to_drop if col in data.columns]

data = data.drop(existing_cols_to_drop, axis=1)


print(data.columns)
```

```python
from sqlalchemy import create_engine
#Exporting data to SQL 
engine = create_engine("postgresql+psycopg2://postgres:Afhaam04@localhost:5432/retail_db")
conn = engine.connect()

rows_inserted = data.to_sql('df_orders', con=conn, index=False, if_exists='append')
print(f"{rows_inserted} rows inserted successfully!")
```
